## functions with variables 


```python
student_name = 'shaurya'
student_age = 18
student_collage = 'bennett, university'
student_course = 'btech, CSE'

print("student name:", student_name)
print("student age:", student_age)
print("student collage:", student_collage)
print("student course:", student_course)

marks_english = 94
marks_maths = 47
marks_physics = 63

def calculate_total(marks_english, marks_maths, marks_physics):
    total = marks_english + marks_maths + marks_physics 
    return total
result = calculate_total(marks_english, marks_maths, marks_physics)
print(result)

```
