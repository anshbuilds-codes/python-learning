# list comprehensions

```python
my_list = [1,2,3,4,5,6]
[2*item for item in my_list]
```

## list comprehension with filter 

```python
my_list = list(range(100))
filteredlist = [item for item in my_list if item % 10 == 0]
filteredlist
```

```python
#we are filtering list by the modulus operator which gives remainder 
#which means when ever and number in ITEM is diveded by number 10 and its remainder is zero save it into filtered list 
```

# dictionaries comprehensions

```python
'''
used to generate dictionaries from unreadable codes :
it is very similer to list comprehensions same as cleaning the code but in dictionaries 
'''
```

```python
animallist = [('a', 'animal'), ('b', 'bear'), ('c', 'cat')]
```

```python
animals = {item[0]: item[1] for item in animallist}
animals
```

```python
# we can also create dictionaries by using key and values and unpacking tuples/list same as done below 
```

```python
animals = {key: value for key, value in animallist}
animals
```
