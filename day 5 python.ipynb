## factorials two types

def factorial(num):
    if type(num) != int:
        return none

    if num < 0:
        return none

    fact = 1
    Counter = 1
    while Counter <= num:
        
        fact = fact * Counter
        Counter = Counter + 1
        
        
    return fact 

factorial(3)
        
    
        
   

```python
def factorial(num):
    if type(num) != int:
        return none

    if num < 0:
        return none
    
    if num == 0:
        return 1

    return num * factorial(num - 1)

factorial(9)
    
```
