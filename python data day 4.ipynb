## creating factorials 

```python
def factorial(num):
    if type(num) != int:
        return None
    if num < 0:
        return None 

    fact = 1
    counter = 1
    while counter <= num:
        fact = fact * counter 
        counter = counter + 1
    return fact 

print(factorial(5))
```

## practicing 

```python
def calculate_total(price, quantity):
    return price * quantity 

calculate_total(3, 100)
```

```python
def discount_calculator(price, discount):
    return price * discount / 100

discount_calculator(1000, 20)
```

```python
maths = 85
physics = 92
chemistry = 71
english = 88
cs = 95

def check_result(marks):
    if marks >= 95:
        return "excellent" 

    if marks >= 80:
        return "good"
    

    if marks >= 60:
        return "pass"
   

    if marks >= 30 :
        return "fail"
    

print("Maths:", check_result(maths))
print("Physics:", check_result(physics))
print("Chemistry:", check_result(chemistry))
print("English:", check_result(english))
print("CS:", check_result(cs))
```
